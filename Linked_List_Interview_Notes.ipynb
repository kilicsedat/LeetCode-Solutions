```
prev → curr → next
```

Everything comes from safely moving these three.

---

# ✅ STEP 1 — Create a Dummy Node (Golden Rule)

```python
dummy = ListNode(0)
dummy.next = head
```

# 🔥 MASTER TEMPLATE

```python
def solve(head):

    dummy = ListNode(0)
    dummy.next = head

    prev = dummy
    curr = head

    while curr:

        next_node = curr.next

        # operation here

        prev = curr
        curr = next_node

    return dummy.next
```

# 🧠 Why Linked List Problems Feel Hard — Cognitive Model

---

## Array Thinking vs Pointer Thinking

Arrays:
```
[3,7,9,2]
```

Linked Lists:
```
node → node → node
```

---

## Brain Conflict

Linked lists require temporal reasoning:

```
What happens AFTER I move this pointer?
```

---

## Hidden Cognitive Load

Track:
- current
- previous
- next
- remaining structure
- modified structure

---

## Expert Trick

Visualize arrows, not values.

```
(node) ──► (node) ──► (node)
```

---

## Three Questions Experts Ask

1. Do I still have access?
2. Did I break structure?
3. Where does traversal continue?

---

## FAANG Cognitive Model

Linked list mastery = safe mutation of a live structure while preserving future access.
